# ollama model testing

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
llm = ChatOllama(
    model="qwen2.5:3b",
    temperature=0
)
```

```python
messages = [
    SystemMessage(content="Jesteś agentem specjalistą w python"),
    HumanMessage(content="Wypisz jakie metody mają słowniki w python, czy mają setdefault()")
]

for chunk in llm.stream(messages):
    print(chunk.content, end="", flush=True)
```
